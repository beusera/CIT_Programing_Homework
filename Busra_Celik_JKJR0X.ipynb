Create a House class that represents a house; attributes:


*   area
*   number of rooms
*   number of bathrooms
*   number of floors
*   has garage
*   has garden
*   any other attributes that you think is important



instance methods:

*   price calculation

There should be at least 1 default parameter

(4 points)

```python
class House:
    def __init__(self, area, rooms, bathrooms, balcony=0, floors=1, garage=False, garden=False):
        self.area = area
        self.rooms = rooms
        self.bathrooms = bathrooms
        self.floors = floors
        self.garage = garage
        self.garden = garden
        self.balcony = balcony

    def price_calculation(self):
        price = self.area * 1200

        if self.garage:
            price += 10000

        if self.garden:
            price += 15000

        return price
```

```python
class Person:
    def __init__(self, name, age, married, children, cars, money):
        self.name = name
        self.age = age
        self.married = married
        self.children = children
        self.cars = cars
        self.money = money

        if married:
          self.family_members = children + 2
        else:
          self.family_members = 1

    def buying_house(self, house_price):  #instance method
        if self.money >= house_price:
            return True
        else:
            return False

```

###A) (5 points)
 Create an Animal class and choose 3 animals that have some things in common (characteristics and behaviour)!

Write the Parent class and the derived classes of the species!

The Child classes should inherit some attributes and override some methods of the Parent class, but should also extend the state with own attributes in each subclass and extend the behaviour with own methods.

Decide yourself:

which methods belong to parent


which should be overridden


which attributes are shared. Explain your choice.

```python
class Animal:
    def __init__(self, name, age, color):
        self.name = name
        self.age = age
        self.color = color

    def eat(self):
        print(self.name, "is eating.")

    def produce_milk(self):
        print(self.name, "can produce milk.")


class Goat(Animal):
    def __init__(self, name, age, color, milk_per_day):
        super().__init__(name, age, color)
        self.milk_per_day = milk_per_day

    def produce_milk(self):
        print(self.name, "produces", self.milk_per_day, "liters of goat milk.")

    def climb(self):
        print(self.name, "is climbing rocks.")


class Cow(Animal):
    def __init__(self, name, age, color, milk_per_day):
        super().__init__(name, age, color)
        self.milk_per_day = milk_per_day

    def produce_milk(self):
        print(self.name, "produces", self.milk_per_day, "liters of cow milk.")

    def graze(self):
        print(self.name, "is eating grass in the field.")


class Sheep(Animal):
    def __init__(self, name, age, color, milk_per_day):
        super().__init__(name, age, color)
        self.milk_per_day = milk_per_day

    def produce_milk(self):
        print(self.name, "produces", self.milk_per_day, "liters of sheep milk.")

    def grow_wool(self):
        print(self.name, "is growing wool.")


goat = Goat("Goat", 3, "white", 4)
cow = Cow("Cow", 5, "black-white", 20)
sheep = Sheep("Sheep", 1, "white", 2)

goat.eat()
goat.produce_milk()
goat.climb()

cow.eat()
cow.produce_milk()
cow.graze()

sheep.eat()
sheep.produce_milk()
sheep.grow_wool()
```

I created an Animal parent class because goats, cows and sheep have common characteristics. The shared attributes are name, age and color because all animals have these features. The eat() method belongs to the parent class because every animal eats.

The child classes inherit these common attributes and methods from the Animal class. I overrode the produce_milk() method because each animal produces a different type and amount of milk.

Each subclass also has its own behaviour:
- Goat can climb rocks
- Cow can graze in the field
- Sheep can grow wool

This is why each subclass has its own additional method.

###B) (6 points)

Create classes and use inheritance to calculate the area of different shapes (specifically, circle and square, you can add triangle or anything else if you want).

Requirements:


1.	Create a base class called Shape that has a method named area().

This method should not have any implementation
2. Create at least two subclasses:

Circle — which takes a radius as a parameter

Square — which takes a side as a parameter

write the area-calculating function for both (or for all)

3. Create objects of both Circle and Square, and print their areas.

Extra: Store these objects in a list and use a loop to print the area of each shape.

```python
import math

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return (self.base * self.height) / 2


circle = Circle(5)
square = Square(4)
triangle = Triangle(6, 3)

print("Circle Area:", circle.area())
print("Square Area:", square.area())
print("Triangle Area:", triangle.area())


shapes = [circle, square, triangle]

for shape in shapes:
    print("Area:", shape.area())
```

Create a list containing only the vowels from all words.

```python
words = ["apple", "sky", "education", "why"]
```

```python
vowels = []

for word in words:
    for letter in word:
        if letter in "aeiou":
            vowels.append(letter)

print(vowels)
```

```python
letters = {}

for word in text:
    for letter in word:
        if letter in letters:
            letters[letter] += 1
        else:
            letters[letter] = 1

print(letters)
```
